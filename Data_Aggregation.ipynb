16. For each branch, calculate the total, average, minimum, maximum and standard deviation of
units sold using a single agg() operation.

```python
result = df.groupby("branch")["units_sold"].agg(
    ["sum", "mean", "min", "max", "std"]
)

print(result)
```

Q17. Calculate total units sold, average discount, maximum customer rating and minimum delivery time for each product.

```python
result = df.groupby("product").agg({
    "units_sold": "sum",
    "discount_pct": "mean",
    "customer_rating": "max",
    "delivery_minutes": "min"
})

print(result)
```

18. For each branch, calculate the total number of sales records using size() and the number of
non-missing customer ratings using count(). Compare the two results.

```python
result = df.groupby("branch").agg(
    total_records=("record_id", "size"),
    customer_ratings=("customer_rating", "count")
)

print(result)
```

19. Define a user-defined aggregation function to calculate the range (maximum - minimum) of units
sold for each product.

```python
def calculate_range(x):
    return x.max() - x.min()

result = df.groupby("product")["units_sold"].agg(calculate_range)

print(result)
```

20. Group the dataset by branch and product, calculate the total units sold and total target units, and
return the result as a DataFrame using as_index=False.

```python
result = df.groupby(
    ["branch", "product"],
    as_index=False
).agg({
    "units_sold": "sum",
    "target_units": "sum"
})

print(result)
```

23. Create a hierarchical index using the branch and product columns. Perform aggregation using
the branch index level to calculate the total units sold by each branch.

```python
result = df.groupby(
    ["branch", "product"]
)["units_sold"].sum()

print("Hierarchical Index:")
print(result)

print("Total Units by Branch:")
print(result.groupby(level="branch").sum())
```

24. Create a pivot table displaying the total units sold, with branches as rows and products as
columns. Include row totals and column totals.

```python
result = pd.pivot_table(
    df,
    values="units_sold",
    index="branch",
    columns="product",
    aggfunc="sum",
    fill_value=0,
    margins=True
)

print(result)
```

26. Identify missing values in customer_rating and replace them with the average customer rating of
the corresponding product within the same branch using groupby() and transform(). Display the
dataset before and after replacement.

```python
print("Before Replacement:")
print(df)

df["customer_rating"] = df["customer_rating"].fillna(
    df.groupby(["branch", "product"])["customer_rating"].transform("mean")
)

print("After Replacement:")
print(df)
```

27. Calculate the weighted average discount percentage for each branch, considering units_sold as
the weight. Use a user-defined function with groupby() and apply(). Calculate the weighted
average as the sum of (units sold × discount percentage) divided by the total units sold.

```python
def weighted_average(group):
    return (
        (group["units_sold"] * group["discount_pct"]).sum()
        / group["units_sold"].sum()
    )

result = df.groupby("branch")[["units_sold", "discount_pct"]].apply(
    weighted_average
)

print(result)
```

28. Identify the top three sales records with the highest units sold for every branch using groupby()
and apply(). Display the branch, product, date and units sold for the selected records.

```python
result = df.groupby(
    "branch",
    group_keys=False
).apply(
    lambda x: x.nlargest(3, "units_sold")
)

print(result[["branch", "product", "day", "units_sold"]])
```

29. Calculate the percentage contribution of each sales record to the total units sold by its respective
branch using transform(). Create a new column named sales_contribution_pct and verify that
the total contribution for each branch is 100% (subject to rounding).

```python
df["sales_contribution_pct"] = (
    df["units_sold"] /
    df.groupby("branch")["units_sold"].transform("sum")
) * 100

print(df)

result = df.groupby("branch")["sales_contribution_pct"].sum()

print("Total Contribution:")
print(result)
```

30. Divide all sales records into four quartile-based categories using pd.qcut() on units_sold: Low
Sales, Below Average Sales, Above Average Sales and High Sales. Group by these categories
and calculate total units sold, average discount percentage, average customer rating, average
delivery time and number of sales records.

```python
df["sales_category"] = pd.qcut(
    df["units_sold"],
    q=4,
    labels=[
        "Low Sales",
        "Below Average Sales",
        "Above Average Sales",
        "High Sales"
    ]
)

result = df.groupby("sales_category", observed=False).agg(
    total_units_sold=("units_sold", "sum"),
    average_discount=("discount_pct", "mean"),
    average_rating=("customer_rating", "mean"),
    average_delivery_time=("delivery_minutes", "mean"),
    number_of_records=("record_id", "size")
)

print(result)
```
