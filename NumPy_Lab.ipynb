## Question 39

Student Marks Analysis: Create a NumPy array containing marks of 10 students in 3 subjects. Calculate each student's total and average marks. Identify students whose average is ≥ 50 using np.where().

```python
import numpy as np

marks = np.array([
    [78, 65, 80], [45, 55, 50], [90, 88, 95], [32, 40, 38],
    [60, 70, 65], [49, 52, 55], [85, 90, 87], [40, 35, 45],
    [72, 68, 75], [55, 58, 60]
])
totals = marks.sum(axis=1)
averages = marks.mean(axis=1)
qualified = np.where(averages >= 50)[0] + 1
print("Marks:\n", marks)
print("Totals:", totals)
print("Averages:", averages)
print("Student numbers with average >= 50:", qualified)
```

## Question 40

Matrix Analysis: Create a 4 × 4 matrix containing numbers 1–16. Find its transpose, row-wise sums, column-wise sums, mean, maximum, and minimum. Extract the central 2 × 2 matrix.

```python
import numpy as np

matrix = np.arange(1, 17).reshape(4, 4)
print("Matrix:\n", matrix)
print("Transpose:\n", matrix.T)
print("Row sums:", matrix.sum(axis=1))
print("Column sums:", matrix.sum(axis=0))
print("Mean:", matrix.mean())
print("Maximum:", matrix.max())
print("Minimum:", matrix.min())
print("Central 2 x 2:\n", matrix[1:3, 1:3])
```

## Question 42

Matrix Operations Challenge: Create two 3 × 3 matrices and perform element-wise addition, element-wise multiplication, matrix multiplication, transpose of the first matrix, and maximum/minimum of the resulting matrix.

```python
import numpy as np

a = np.arange(1, 10).reshape(3, 3)
b = np.arange(9, 0, -1).reshape(3, 3)
addition = a + b
elementwise = a * b
product = a @ b
print("A:\n", a)
print("B:\n", b)
print("Element-wise addition:\n", addition)
print("Element-wise multiplication:\n", elementwise)
print("Matrix multiplication:\n", product)
print("Transpose of A:\n", a.T)
print("Maximum in matrix product:", product.max())
print("Minimum in matrix product:", product.min())
```

## Question 43

Array Manipulation Challenge: Create an array containing 1–24, reshape it into a 4 × 6 matrix, extract selected rows and columns, transpose it, calculate row/column statistics, and identify unique values.

```python
import numpy as np

arr = np.arange(1, 25)
matrix = arr.reshape(4, 6)
print("4 x 6 matrix:\n", matrix)
print("Selected rows (rows 1 and 3):\n", matrix[[0, 2], :])
print("Selected columns (columns 2 and 5):\n", matrix[:, [1, 4]])
print("Transpose:\n", matrix.T)
print("Row means:", matrix.mean(axis=1))
print("Column sums:", matrix.sum(axis=0))
print("Unique values:", np.unique(matrix))
```
