Q1. Read the CSV file into a DataFrame called df.

```python
import pandas as pd

df = pd.read_csv('Retail_Data_Science_Dataset.csv')
```

Q2. Find the total number of rows and columns and display the first 5 rows.

```python
print(df.shape)

print(df.head())
```

Q3. Extract only product, units_sold and unit_price columns.

```python
print(df[['product', 'units_sold', 'unit_price']])
```

Q4. Find summary statistics (mean, standard deviation, min, max, quartiles).

```python
print(df[['units_sold', 'unit_price']].describe())
```

Q5. Select the first 10 rows and first 4 columns using iloc.

```python
print(df.iloc[0:10, 0:4])
```

Q6. Display only rows where branch equals 'Mathura'.

```python
print(df[df['branch'] == 'Mathura'])
```

Q7. Find rows where units_sold is greater than 30 and discount_pct is greater than 0.

```python
print(df[(df['units_sold'] > 30) & (df['discount_pct'] > 0)])
```

Q8. Count total transaction records for each distinct product.

```python
print(df['product'].value_counts())
```

Q9. Sort the entire DataFrame by customer_rating in descending order.

```python
print(df.sort_values(by='customer_rating', ascending=False))
```

Q10. Detect and count missing (NaN) values in every column.

```python
print(df.isnull().sum())
```
